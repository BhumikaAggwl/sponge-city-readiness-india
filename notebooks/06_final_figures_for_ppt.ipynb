```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px

plt.rcParams["figure.dpi"] = 300
plt.rcParams["savefig.dpi"] = 300
```

```python
df = pd.read_csv(
    "../data/processed/scri_v2_final.csv"
)

df.head()
```

SCRI DISTRIBUTION

```python

plt.figure(figsize=(8,5))

plt.hist(
    df["scri"],
    bins=30
)

plt.xlabel("SCRI")

plt.ylabel("Count")

plt.title(
    "Distribution of Sponge City Readiness Index"
)

plt.tight_layout()

plt.savefig(
    "../outputs/figures/01_scri_distribution.png"
)

plt.show()
```

```python
df["scri_class"] \
.value_counts() \
.plot(
    kind="bar"
)

plt.title(
    "SCRI Readiness Classes"
)

plt.tight_layout()

plt.savefig(
    "../outputs/figures/03_scri_classes.png"
)

plt.show()
```

```python
priority = df.nsmallest(
    100,
    "scri"
)

plt.figure(
    figsize=(8,6)
)

plt.scatter(

    priority["lon"],

    priority["lat"],

    c=priority["scri"]

)

plt.colorbar()

plt.title(
    "Priority Intervention Zones"
)

plt.tight_layout()

plt.savefig(
    "../outputs/figures/04_priority_zones.png"
)

plt.show()
```

```python
drivers = [

    "rainfall_score",

    "soil_score",

    "terrain_score",

    "water_norm",

    "permeability_norm",

    "vegetation_proxy"

]

driver_means = (
    df[drivers]
    .mean()
    .sort_values(
        ascending=False
    )
)

driver_means.plot(
    kind="bar"
)

plt.title(
    "Average SCRI Components"
)

plt.tight_layout()

plt.savefig(
    "../outputs/figures/05_scri_drivers.png"
)

plt.show()
```
